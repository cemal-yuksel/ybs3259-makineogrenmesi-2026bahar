    <p style="color:#bbf7d0; font-size:0.88em; margin:0;">Birbirine <strong>benzer veri noktalarından</strong> oluşan grup. Her nokta yalnızca bir kümeye aittir.</p>

  <div style="flex:1; min-width:200px; background:#0c1a3b; border-radius:12px; padding:18px; text-align:center;">
    <div style="font-size:2em; margin-bottom:8px;">🔢</div>
    <h4 style="color:#93c5fd; margin:0 0 6px 0;">K Değeri</h4>
    <p style="color:#bfdbfe; font-size:0.88em; margin:0;">Oluşturulacak <strong>küme sayısı</strong>. Kullanıcı tarafından önceden belirlenir. Doğru K seçimi kritiktir!</p>
  </div>

  <div style="flex:1; min-width:200px; background:#1c1917; border-radius:12px; padding:18px; text-align:center;">
    <div style="font-size:2em; margin-bottom:8px;">📐</div>
    <h4 style="color:#fde68a; margin:0 0 6px 0;">Öklid Mesafesi</h4>
    <p style="color:#fef3c7; font-size:0.88em; margin:0;">İki nokta arasındaki <strong>düz çizgi (kuş uçuşu) mesafesi</strong>. K-Means'in varsayılan uzaklık ölçütüdür.</p>

  <div style="flex:1; min-width:200px; background:#0f2a1f; border-radius:12px; padding:18px; text-align:center;">

    <p style="color:#a7f3d0; font-size:0.88em; margin:0;">Algoritmanın <strong>bir çevrim tamamlaması</strong>. Atama + Güncelleme adımlarının her tekrarına bir iterasyon denir.</p>
  </div>

  <h4 style="color:#fbbf24; margin-top:0;">📐 MinMaxScaler Formülü</h4>
  <div style="background:#0f172a; border-radius:8px; padding:12px 18px; text-align:center; font-size:1.1em; color:#e2e8f0; font-family:monospace; letter-spacing:1px;">
    X_scaled = (X − X_min) / (X_max − X_min)
  </div>
  <p style="color:#94a3b8; font-size:0.9em; margin:10px 0 0;">
    &bull; Sonuç her zaman <strong style="color:#7dd3fc;">0 ile 1 arasında</strong> olur<br>
    &bull; <code style="color:#a5b4fc;">fit_transform()</code> → eğitim verisine uygula + dönüştür<br>
    &bull; <code style="color:#a5b4fc;">transform()</code> → test verisine sadece dönüştür (fit etme — data leakage önlenir!)
  </p>
</div>

```python
# ─────────────────────────────────────────────────────────
# ✂️ EĞİTİM / TEST VERİSİNE BÖLME
# ─────────────────────────────────────────────────────────
# NOT: K-Means denetimsiz bir algoritma olduğundan y (etiket) yoktur.
# Bu yüzden sadece X'i (özellikler) böleceğiz.

from sklearn.model_selection import train_test_split

# %70 eğitim, %30 test — random_state tekrarlanabilirlik için sabitlendi
X_train, X_test = train_test_split(df, test_size=0.3, random_state=15)

print(f"📐 Toplam veri   : {len(df)} satır")
print(f"🟢 Eğitim seti   : {len(X_train)} satır  (%{100*len(X_train)//len(df)})")
print(f"🔵 Test seti     : {len(X_test)} satır  (%{100*len(X_test)//len(df)})")
print()
print("💡 Neden train-test split yapıyoruz?")
print("   → Modelimizin görmediği veriye (test) ne kadar genelleyebildiğini ölçmek için.")
```

```python
# ─────────────────────────────────────────────────────────
# 📏 MINMAXSCALer İLE NORMALİZASYON
# ─────────────────────────────────────────────────────────
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()   # Varsayılan: aralık [0, 1]

# fit_transform → eğitim verisindeki min/max öğren, sonra dönüştür
X_train_scaled = scaler.fit_transform(X_train)

# transform → sadece dönüştür (min/max bilgisi eğitimden alındı)
# ⚠️ Test verisine fit_transform kullanmak DATA LEAKAGE'a yol açar!
X_test_scaled  = scaler.transform(X_test)

print("✅ Normalleştirme tamamlandı!")
print()
print(f"{'':25} {'Eğitim':>12} {'Test':>12}")
print("─" * 52)
print(f"{'Ölçeklenmemiş Min':25} {X_train.min().min():>12.3f} {X_test.min().min():>12.3f}")
print(f"{'Ölçeklenmemiş Max':25} {X_train.max().max():>12.3f} {X_test.max().max():>12.3f}")
print("─" * 52)
print(f"{'Ölçeklenmiş Min':25} {X_train_scaled.min():>12.3f} {X_test_scaled.min():>12.3f}")
print(f"{'Ölçeklenmiş Max':25} {X_train_scaled.max():>12.3f} {X_test_scaled.max():>12.3f}")
```

```python
# ─────────────────────────────────────────────────────────
# 🔁 ÖNCE vs SONRA — Ölçeklemenin Etkisini Görelim
# ─────────────────────────────────────────────────────────

fig, axes = plt.subplots(1, 2, figsize=(15, 6))
fig.suptitle("Ölçekleme Öncesi vs Sonrası", fontsize=15, fontweight='bold')

# --- ÖNCE (Ham veri) ---
axes[0].scatter(X_train['Annual_Income'], X_train['Spending_Score'],
                c='#6366f1', alpha=0.6, edgecolors='white', s=70, linewidths=0.4)
axes[0].set_title('Ölçekleme ÖNCESİ\n(Ham veri — farklı aralıklar)', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Annual_Income')
axes[0].set_ylabel('Spending_Score')
axes[0].set_facecolor('#f8fafc')

# X ve Y eksenlerini karşılaştırmayı kolaylaştırmak için sınırları göster
x_range = X_train['Annual_Income'].max() - X_train['Annual_Income'].min()
y_range = X_train['Spending_Score'].max() - X_train['Spending_Score'].min()
axes[0].text(0.03, 0.95, f"X aralığı: {x_range:.1f}\nY aralığı: {y_range:.1f}",
             transform=axes[0].transAxes, fontsize=10,
             verticalalignment='top', color='#b91c1c',
             bbox=dict(boxstyle='round', facecolor='#fee2e2', alpha=0.8))

# --- SONRA (Ölçeklenmiş veri) ---
axes[1].scatter(X_train_scaled[:, 0], X_train_scaled[:, 1],
                c='#22c55e', alpha=0.6, edgecolors='white', s=70, linewidths=0.4)
axes[1].set_title('Ölçekleme SONRASI\n(MinMaxScaler — 0 ile 1 arasında)', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Annual_Income (ölçeklenmiş)')
axes[1].set_ylabel('Spending_Score (ölçeklenmiş)')
axes[1].set_facecolor('#f0fdf4')

axes[1].text(0.03, 0.95, "Her iki eksen: [0, 1]\nEşit ağırlık ✅",
             transform=axes[1].transAxes, fontsize=10,
             verticalalignment='top', color='#15803d',
             bbox=dict(boxstyle='round', facecolor='#dcfce7', alpha=0.8))

plt.tight_layout()
plt.show()

print("📌 Grafiklerin şekli aynı — ölçekleme veri yapısını bozmaz, sadece aralığı değiştirir!")
```

```python
# ─────────────────────────────────────────────────────────
# 📐 ELBOW METODU — WCSS DEĞERLERİNİ HESAPLA
# ─────────────────────────────────────────────────────────
from sklearn.cluster import KMeans

wcss = []   # Her K için WCSS değerini saklayacağız

# K=1'den K=10'a kadar her değer için model kurup WCSS'i kaydet
for k in range(1, 11):
    kmeans_temp = KMeans(
        n_clusters=k,
        init='k-means++',   # Akıllı başlangıç noktası seçimi
        n_init=10,          # 10 farklı başlangıçtan en iyisini al
        random_state=42     # Tekrarlanabilirlik için sabit tohum
    )
    kmeans_temp.fit(X_train_scaled)
    wcss.append(kmeans_temp.inertia_)  # inertia_ = WCSS değeri

# Sonuçları tablolu yazdır
print(f"{'K':>4} | {'WCSS (Inertia)':>18} | {'Öncekine Göre Düşüş':>20}")
print("─" * 48)
for i, w in enumerate(wcss):
    dusu = f"↓ {wcss[i-1]-w:.2f}" if i > 0 else "  (başlangıç)"
    print(f"{i+1:>4} | {w:>18.4f} | {dusu:>20}")
```

  <p style="color:#92400e; font-size:0.93em; line-height:1.7; margin:0;">
    WCSS, K artıkça <strong>her zaman düşer</strong> — bu normal. Soru şu: <em>Düşüşün en fazla yavaşladığı K hangisi?</em>
    <br><br>
    🔴 K=1 → 2: <strong>Çok büyük düşüş</strong> (mantıklı, 1 kümeden 2'ye geçiş büyük iyileşme)<br>
    🟡 K=2 → 3: <strong>Önemli düşüş</strong> — dirsek burada!<br>
    🟢 K=3 → 4+: <strong>Düşüş yavaşlar</strong> — daha fazla küme anlamlı katkı sağlamaz
    <br><br>
    📌 <strong>Kural:</strong> Dirsek noktasından sonra WCSS düşüşü marjinal kalır → o nokta optimal K'dır.
  </p>
</div>

```python
# ─────────────────────────────────────────────────────────
# 🦾 KNEELOCATOR — Dirseği OTOMATİK Bul
# ─────────────────────────────────────────────────────────
# Elbow grafiğini gözle yorumlamak zor olabilir.
# kneed kütüphanesi matematiksel olarak dirsek noktasını bulur!

# Kurulum (daha önce yüklenmemişse):
# !pip install kneed

try:
    from kneed import KneeLocator
except ModuleNotFoundError:
    import sys
    import subprocess
    subprocess.check_call([sys.executable, "-m", "pip", "install", "kneed", "-q"])
    from kneed import KneeLocator

kl = KneeLocator(
    x=range(1, 11),   # K değerleri
    y=wcss,           # WCSS değerleri
    curve='convex',   # Eğri tipi: dışbükey (convex) — azalan WCSS için
    direction='decreasing'  # Yön: azalan
)

print(f"🎯 KneeLocator'ın önerdiği optimal K değeri: K = {kl.elbow}")
print()
print("📌 Bu değer, Elbow grafiğindeki dirsek noktasına karşılık gelir.")
print("   Kütüphane bu hesabı otomatik ve objektif olarak yapar — göz yanılması yok!")
```

```python
# ─────────────────────────────────────────────────────────
# 🤖 FINAL MODEL — K=3 İLE K-MEANS EĞİTİMİ
# ─────────────────────────────────────────────────────────
# Elbow ve KneeLocator K=3 önerdi → şimdi asıl modelimizi kuruyoruz!

optimal_k = kl.elbow   # KneeLocator'dan gelen K değeri

kmeans = KMeans(
    n_clusters=optimal_k,   # 3 küme oluştur
    init='k-means++',       # Akıllı başlangıç (düz rastgele değil)
    n_init=10,              # 10 farklı başlangıçtan en iyisini seç
    random_state=42         # Tekrarlanabilirlik
)

# Modeli EĞİTİM verisi üzerinde eğit
kmeans.fit(X_train_scaled)

print(f"✅ K-Means modeli K={optimal_k} ile başarıyla eğitildi!")
print()
print(f"📊 Model Bilgileri:")
print(f"   • Küme sayısı      : {kmeans.n_clusters}")
print(f"   • İterasyon sayısı : {kmeans.n_iter_}  (yakınsama burada sağlandı)")
print(f"   • Final WCSS       : {kmeans.inertia_:.4f}")
print()
print(f"📍 Centroid Koordinatları (ölçeklenmiş uzayda):")
for i, c in enumerate(kmeans.cluster_centers_):
    print(f"   Küme {i} merkezi → Annual_Income: {c[0]:.4f} | Spending_Score: {c[1]:.4f}")
```

```python
# ─────────────────────────────────────────────────────────
# 🔮 TAHMİN — Test Verisi Hangi Kümelere Ait?
# ─────────────────────────────────────────────────────────
# Model eğitim verisiyle öğrendi.
# Şimdi görmediği test verisi için küme tahmini yapıyor.

y_pred = kmeans.predict(X_test_scaled)

print("🔮 Test verisi için küme tahminleri:")
print()
print(f"   Toplam test noktası : {len(y_pred)}")
import numpy as np
for k in range(optimal_k):
    sayi = np.sum(y_pred == k)
    yuzde = 100 * sayi / len(y_pred)
    bar = '█' * int(yuzde / 3)
    print(f"   Küme {k}: {sayi:>3} müşteri  ({yuzde:.1f}%)  {bar}")

print()
print("📌 Her test noktası, eğitim aşamasında öğrenilen")
print("   centroid'lara olan mesafesine göre atandı.")
```

```python
# ─────────────────────────────────────────────────────────
# 📊 K=2..10 İÇİN SİLHOUETTE SCORE HESAPLA
# ─────────────────────────────────────────────────────────
# silhouette_score: küme içi ve kümeler arası mesafeleri karşılaştırır.
# Elbow'daki gibi K=2'den başlıyoruz (K=1 silhouette tanımsız!)

from sklearn.metrics import silhouette_score

sil_skorlar = []     # Her K için silhouette score listesi
k_aralik    = range(2, 11)   # K=2,3,...,10

for k in k_aralik:
    km_tmp = KMeans(n_clusters=k, init='k-means++',
                    n_init=10, random_state=42)
    etiket = km_tmp.fit_predict(X_train_scaled)
    skor   = silhouette_score(X_train_scaled, etiket)
    sil_skorlar.append(skor)

# ── Tablo formatında göster ──
print(f"{'K':>4}  {'Silhouette Score':>17}  {'Değerlendirme'}")
print("─" * 42)
for k, s in zip(k_aralik, sil_skorlar):
    if   s >= 0.70: yorum = "🌟 Mükemmel"
    elif s >= 0.50: yorum = "✅ İyi"
    elif s >= 0.25: yorum = "⚠️  Orta"
    else:           yorum = "❌ Zayıf"
    isaretli = " ◄ OPTİMAL" if k == optimal_k else ""
    print(f"{k:>4}  {s:>17.4f}  {yorum}{isaretli}")
```

<div style="background:#1c1917; border-left:5px solid #fb923c; border-radius:10px; padding:18px 22px; margin:12px 0;">
  <h4 style="color:#fed7aa; margin-top:0;">🔶 Davies-Bouldin Index (DBI) Nedir?</h4>
  <p style="color:#e7d0c0; font-size:0.91em; line-height:1.75; margin:0 0 10px 0;">
    Silhouette'e alternatif bir değerlendirme metriğidir. Her küme çifti için şunu sorar:<br>
    <em>"İki kümenin içi dağılımı, aralarındaki mesafeye göre ne kadar büyük?"</em>
  </p>
  <p style="color:#fdba74; font-size:0.9em; margin:0 0 10px 0;">
    <code style="background:#292524; color:#fb923c; padding:4px 8px; border-radius:4px;">DBI = ortalama( max_j≠i ( (σ_i + σ_j) / d(c_i, c_j) ) )</code>
  </p>
  <ul style="color:#e7d0c0; font-size:0.89em; line-height:1.8; margin:0; padding-left:16px;">
    <li><strong style="color:#fdba74;">σ_i</strong> = Küme i'nin ortalama içi yarıçapı (ne kadar sıkışık?)</li>
    <li><strong style="color:#fdba74;">d(c_i, c_j)</strong> = İki küme merkezi arasındaki mesafe</li>
    <li>DBI <strong style="color:#4ade80;">düşüklüğü</strong> iyidir → Kümeler birbirinden uzak ve içte sıkışık</li>
    <li>Silhouette'ten farkı: <strong>değer aralığı 0'dan ∞'a</strong> — 0'a yakın = mükemmel</li>
  </ul>
</div>

```python
# ─────────────────────────────────────────────────────────
# 🔶 K=2..10 İÇİN DAVIES-BOULDIN INDEX HESAPLA
# ─────────────────────────────────────────────────────────
# Düşük DBI = kümeler arası iyi ayrım + içi düşük dağılım → ideal

from sklearn.metrics import davies_bouldin_score

dbi_skorlar = []

for k in k_aralik:
    km_tmp  = KMeans(n_clusters=k, init='k-means++',
                     n_init=10, random_state=42)
    etiket  = km_tmp.fit_predict(X_train_scaled)
    dbi     = davies_bouldin_score(X_train_scaled, etiket)
    dbi_skorlar.append(dbi)

# ── Tablo ──
print(f"{'K':>4}  {'DBI':>8}  {'Değerlendirme'}")
print("─" * 38)
for k, d in zip(k_aralik, dbi_skorlar):
    if   d < 0.5:  yorum = "🌟 Mükemmel"
    elif d < 0.8:  yorum = "✅ İyi"
    elif d < 1.2:  yorum = "⚠️  Orta"
    else:          yorum = "❌ Zayıf"
    isaretli = " ◄ OPTİMAL" if k == optimal_k else ""
    print(f"{k:>4}  {d:>8.4f}  {yorum}{isaretli}")

en_iyi_dbi_k = list(k_aralik)[dbi_skorlar.index(min(dbi_skorlar))]
print(f"\n🏆 En düşük DBI (en iyi) → K={en_iyi_dbi_k}  ({min(dbi_skorlar):.4f})")
```

```python
# ─────────────────────────────────────────────────────────
# 📊 ÜÇ METRİK YAN YANA — Karşılaştırma Grafiği
# ─────────────────────────────────────────────────────────
# WCSS, Silhouette Score ve DBI'yi tek grafikte gösteriyoruz.
# Her metriğin en iyi K'sı görsel olarak işaretleniyor.

import numpy as np

wcss_norm = np.array(wcss[1:]) / max(wcss[1:])   # K=2..10 için normalize WCSS
sil_arr   = np.array(sil_skorlar)
dbi_norm  = np.array(dbi_skorlar) / max(dbi_skorlar)   # normalize DBI

fig, axes = plt.subplots(1, 3, figsize=(16, 5))
fig.patch.set_facecolor('#0f172a')

grafik_veri = [
    (axes[0], wcss_norm,  '#60a5fa', 'WCSS (normalize)',  'Küçük = İyi', False),
    (axes[1], sil_arr,    '#4ade80', 'Silhouette Score', 'Büyük = İyi', True),
    (axes[2], dbi_norm,   '#fb923c', 'DBI (normalize)',   'Küçük = İyi', False),
]

for ax, veri, renk, baslik, ipucu, buyuk_iyi in grafik_veri:
    ax.plot(list(k_aralik), veri, 'o-', color=renk,
            linewidth=2, markersize=8, markerfacecolor='white')
    # Optimal noktayı vurgula
    idx = list(k_aralik).index(optimal_k)
    ax.scatter([optimal_k], [veri[idx]], s=200, color='#f59e0b',
               zorder=5, label=f'K={optimal_k}')
    ax.set_title(baslik, fontsize=12, fontweight='bold', color='#e2e8f0', pad=10)
    ax.set_xlabel('K', fontsize=11, color='#94a3b8')
    ax.set_ylabel(ipucu, fontsize=10, color=renk)
    ax.set_xticks(list(k_aralik))
    ax.tick_params(colors='#94a3b8')
    ax.set_facecolor('#1e293b')
    for spine in ax.spines.values():
        spine.set_edgecolor('#334155')
    ax.legend(fontsize=10, labelcolor='#fbbf24',
              facecolor='#0f172a', edgecolor='#334155')

plt.suptitle('Üç Metrik ile K Seçimi Karşılaştırması',
             fontsize=14, fontweight='bold', color='#e2e8f0', y=1.02)
plt.tight_layout()
plt.show()

print("\n📌 Üç metriğin de K=", optimal_k, "i önerdiği görülüyor.")
print("   → WCSS düşüyor  |  Silhouette en yüksek  |  DBI en düşük")
```

```python
# ─────────────────────────────────────────────────────────
# 🗂️ TÜM VERİYİ KÜMELERE ATA
# ─────────────────────────────────────────────────────────
# Modeli eğitim verisinde eğittik, test verisiyle doğruladık.
# Şimdi TÜM müşterilere küme etiketi atıyoruz (tam tablo için).

import numpy as np

# Tüm veriyi ölçekle (scaler X_train'e göre fit edilmişti)
X_tum        = df.values                            # numpy array
X_tum_scaled = scaler.transform(X_tum)             # normalize

# Tüm veri için küme tahminleri
labels_tum   = kmeans.predict(X_tum_scaled)

# Orijinal df'ye etiket ekle (kopyasına)
df_sonuc = df.copy()
df_sonuc['Kume'] = labels_tum

# Centroid'ları orijinal ölçeğe geri dönüştür
centroid_orijinal = scaler.inverse_transform(kmeans.cluster_centers_)

print("✅ Tüm müşterilere küme etiketi atandı!")
print()
print(f"{'Küme':>6}  {'Müşteri Sayısı':>15}  {'Annual_Income (ort)':>20}  {'Spending_Score (ort)':>21}")
print("─" * 70)
for k in range(optimal_k):
    alt_kume = df_sonuc[df_sonuc['Kume'] == k]
    print(f"  {k:>4}  {len(alt_kume):>15}  {alt_kume['Annual_Income'].mean():>20.4f}  {alt_kume['Spending_Score'].mean():>21.4f}")

print()
print("📍 Centroid Koordinatları (orijinal ölçekte):")
for i, c in enumerate(centroid_orijinal):
    print(f"   Küme {i} → Annual_Income: {c[0]:.4f} | Spending_Score: {c[1]:.4f}")
```

```python
# ─────────────────────────────────────────────────────────
# 🗺️ TÜM VERİ — KÜME GÖRSELLEŞTİRMESİ
# ─────────────────────────────────────────────────────────
# Tüm müşterileri küme renklerine göre scatter plot ile gösteriyoruz.
# Centroid'lar altın yıldız ile işaretleniyor.

RENKLER_6  = ['#ef4444', '#3b82f6', '#22c55e', '#f59e0b', '#a855f7', '#06b6d4']
SEGMENT_AD = ['Segment 0', 'Segment 1', 'Segment 2',
              'Segment 3', 'Segment 4', 'Segment 5']

fig, ax = plt.subplots(figsize=(11, 7))
fig.patch.set_facecolor('#0f172a')
ax.set_facecolor('#1e293b')

# Her kümeyi ayrı renkte çiz
for k in range(optimal_k):
    maske = labels_tum == k
    ax.scatter(
        X_tum_scaled[maske, 0],
        X_tum_scaled[maske, 1],
        c=RENKLER_6[k],
        label=f'Küme {k}  (n={maske.sum()})',
        alpha=0.80, s=80,
        edgecolors='white', linewidths=0.5
    )

# Centroid yıldızları
centroidler_k = kmeans.cluster_centers_
ax.scatter(
    centroidler_k[:, 0], centroidler_k[:, 1],
    marker='*', s=450, c='#fbbf24',
    edgecolors='#1e293b', linewidths=1.5,
    zorder=5, label='Centroid ⭐'
)
for i, (cx, cy) in enumerate(centroidler_k):
    ax.annotate(f'  C{i}', (cx, cy), fontsize=12,
                fontweight='bold', color='#fbbf24',
                xytext=(6, 4), textcoords='offset points')

ax.set_title(f'Tüm Müşteriler — K-Means Kümeleme (K={optimal_k})',
             fontsize=14, fontweight='bold', color='#e2e8f0', pad=14)
ax.set_xlabel('Annual_Income (ölçeklenmiş)', fontsize=12, color='#94a3b8')
ax.set_ylabel('Spending_Score (ölçeklenmiş)', fontsize=12, color='#94a3b8')
ax.tick_params(colors='#94a3b8')
for spine in ax.spines.values():
    spine.set_edgecolor('#334155')
ax.legend(fontsize=11, facecolor='#0f172a', edgecolor='#334155',
          labelcolor='#e2e8f0', loc='best')

plt.tight_layout()
plt.show()

print(f"📊 Toplam {len(df_sonuc)} müşteri → {optimal_k} kümeye atandı")
```

```python
# ─────────────────────────────────────────────────────────
# 🏷️ MÜŞTERİ SEGMENTLERİNİ İSİMLENDİR
# ─────────────────────────────────────────────────────────
# Her kümenin centroid değerlerine bakarak anlamlı isimler veriyoruz.
# Annual_Income düşük/orta/yüksek × Spending_Score düşük/orta/yüksek

import pandas as pd

# Centroid'ları orijinal ölçekte incele
centroid_df = pd.DataFrame(
    centroid_orijinal,
    columns=['Annual_Income_ort', 'Spending_Score_ort']
)
centroid_df.index.name = 'Kume'

# Eşik noktaları hesapla (ölçek referansı için)
gelir_median  = df['Annual_Income'].median()
harcama_median = df['Spending_Score'].median()

# Otomatik segment isimlendirme mantığı
segment_isim = {}
for k in range(optimal_k):
    gelir    = centroid_orijinal[k, 0]
    harcama  = centroid_orijinal[k, 1]
    yuksek_g = gelir    > gelir_median
    yuksek_h = harcama  > harcama_median
    if   yuksek_g and     yuksek_h: isim = "💎 Elit Müşteri (Yüksek Gelir & Harcama)"
    elif yuksek_g and not yuksek_h: isim = "💰 Tasarruflu Zengin (Yüksek Gelir, Düşük Harcama)"
    elif not yuksek_g and yuksek_h: isim = "🛍️  Savurgan Müşteri (Düşük Gelir, Yüksek Harcama)"
    else:                            isim = "🏠 Standart Müşteri (Düşük Gelir & Harcama)"
    segment_isim[k] = isim

centroid_df['Segment Adı'] = centroid_df.index.map(segment_isim)

print("🏷️  Müşteri Segmenti İsimlendirmesi")
print("─" * 80)
for k, row in centroid_df.iterrows():
    print(f"\n  Küme {k} → {row['Segment Adı']}")
    print(f"           Annual_Income : {row['Annual_Income_ort']:>8.4f}")
    print(f"           Spending_Score: {row['Spending_Score_ort']:>8.4f}")
    print(f"           Müşteri sayısı: {(df_sonuc['Kume']==k).sum()}")
```
