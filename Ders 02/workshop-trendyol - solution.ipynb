### 🎯 Görev
Veri analizi ve görselleştirme için ihtiyaç duyduğunuz tüm kütüphaneleri import ediniz ve görsel tema ayarlarını yapılandırınız.

**Yapmanız Gerekenler:**
1. `numpy` → sayısal hesaplamalar için `np` takma adıyla
2. `pandas` → veri manipülasyonu için `pd` takma adıyla
3. `matplotlib.pyplot` → statik grafikler için `plt` takma adıyla
4. `seaborn` → istatistiksel görselleştirme için `sns` takma adıyla
5. `plotly.express` → interaktif görselleştirme için `px` takma adıyla
6. **Renk Teması:** Tüm görselleştirmelerde **pastel tonlar** kullanılacak şekilde `sns.set_palette()` ve `plt.rcParams` ile genel seaborn/matplotlib görsel temasını tanımlayınız.

## 🔍 **Bölüm 8 — Eksik Veri Analizi**

<div style="background: #FFF8F0; border-left: 5px solid #FF6000; padding: 15px 20px; border-radius: 0 8px 8px 0; margin: 10px 0;">

### 🎯 **Görev**
Gerçek dünya veri setlerinde eksik veriler kaçınılmazdır. Trendyol gibi yoğun işlem hacmine sahip platformlarda kargaşa anında bazı alanlar boş kalabilir. Bu adımda eksik verileri **tespit** ediniz ve **raporlayınız.**

**Yapmanız Gerekenler:**

**Adım 1 — Toplam eksik değer sayısı:**
- `df.isnull().sum()` ile her sütundaki eksik değer sayısını bulunuz.

**Adım 2 — Eksik veri oranı (%):**
- Her sütundaki eksik değerlerin veri setine oranını yüzde olarak hesaplayınız.
- Formül: `(df.isnull().sum() / len(df)) * 100`

**Adım 3 — Özet tablo oluşturunuz:**
- Aşağıdaki gibi sütun adı, eksik değer sayısı ve oranını içeren bir `DataFrame` oluşturunuz:

```python
eksik_ozet = pd.DataFrame({
    "Eksik Sayi"  : df.isnull().sum(),
    "Eksik Oran %": (df.isnull().sum() / len(df) * 100).round(2)
})
eksik_ozet = eksik_ozet[eksik_ozet["Eksik Sayi"] > 0].sort_values("Eksik Sayi", ascending=False)
```

**Adım 4 — Yorumlayınız:**
- Hangi sütunlar eksik veri içeriyor?
- Bu eksik verilerle ne yapılabilir? (silme, ortalama ile doldurma, medyan ile doldurma…)

> 💡 **İpucu:** `df.isnull()` → her hücre için `True/False` döner. `.sum()` → `True` değerlerini (yani eksikleri) sayar.

> ⚠️ **Veri Kalitesi Notu:** `product_price` sütunundaki eksik değerler doğrudan gelir hesaplamalarını etkiler! Bir e-ticaret analistinin bu durumla başa çıkma stratejisi ne olmalıdır?

> 🧠 **Tartışma:** `seller_rating` boş olan bir sipariş için satıcıya haksızlık mı edilmiş olur? Eksik rating değerlerini sıfır kabul etmek doğru olur mu?

</div>

````python
# Eksik veri yüzdesi
missing = (df.isnull().sum() / len(df) * 100).sort_values(ascending=False)
missing = missing[missing > 0].rename("Eksik Veri (%)")
print(missing.to_string())

# Görselleştirme
fig, ax = plt.subplots(figsize=(8, max(3, len(missing) * 0.5)))
bars = ax.barh(missing.index, missing.values, color=PASTEL_PALETTE[:len(missing)])
ax.set_xlabel("Eksik Veri (%)")
ax.set_title("Sütunlara Göre Eksik Veri Oranı")
ax.set_xlim(0, 105)

for bar, val in zip(bars, missing.values):
    ax.text(val + 0.5, bar.get_y() + bar.get_height() / 2,
            f"%{val:.1f}", va="center", fontsize=9)

plt.tight_layout()
plt.show()
````

### 🎯 **Görev**
`product_category` ve `payment_type` sütunlarını birlikte gruplandırarak karma bir analiz yapınız.

**Yapmanız Gerekenler:**

**Adım 1 —** Her ödeme yöntemi için ortalama sipariş tutarını bulunuz:
```python
df.groupby("payment_type")["product_price"].mean().round(2).sort_values(ascending=False)
```

```python
pivot = df.pivot_table(
    values  = "product_price",
    index   = "product_category",
    columns = "payment_type",
    aggfunc = "mean"

```

**Adım 3 —** `return_status` sütununu kullanarak her kategorideki iade oranını hesaplayınız:
```python

).round(2).sort_values(ascending=False)
```

## 📦 **Bölüm 12 — Aykırı Değer Tespiti: Boxplot Analizi**

<div style="background: #FFF8F0; border-left: 5px solid #FF6000; padding: 15px 20px; border-radius: 0 8px 8px 0; margin: 10px 0;">

### 📋 **Senaryo**
Trendyol Veri Kalitesi Ekibi, makine öğrenmesi modeli kurmadan önce sayısal sütunlardaki **aykırı değerlerin (outlier)** varlığını görsel olarak tespit etmek istiyor. Aykırı değerler; model performansını olumsuz etkileyebilir, hatalı kayıtlara işaret edebilir ya da gerçek anomilik durumları yansıtıyor olabilir.

### 🎯 **Görev**
Veri setindeki sayısal sütunlar için **boxplot (kutu grafiği)** oluşturarak aykırı değerleri görselleştiriniz.

**Yapmanız Gerekenler:**
1. Aşağıdaki 5 sayısal sütun için tek bir figür üzerinde yan yana boxplot çiziniz:
   - `product_price`, `discount_rate`, `delivery_time_days`, `seller_rating`, `product_rating`

2. Her grafiğe başlık, eksen etiketleri ekleyiniz.

3. Pastel ton renk paleti kullanınız (Bölüm 0'da tanımladığınız tema geçerlidir).

```python
sayisal_sutunlar = ["product_price", "discount_rate", "delivery_time_days", "seller_rating", "product_rating"]

fig, axes = plt.subplots(1, 5, figsize=(20, 6))
fig.suptitle("Trendyol — Sayısal Değişkenlerde Aykırı Değer Analizi", 
             fontsize=14, fontweight="bold", y=1.02)

for i, sutun in enumerate(sayisal_sutunlar):
    sns.boxplot(y=df[sutun], ax=axes[i])
    axes[i].set_title(sutun, fontsize=11)
    axes[i].set_xlabel("")

plt.tight_layout()
plt.show()
```

4. **Yorumlayınız:** Hangi sütunlarda belirgin outlier var? Bu aykırı değerleri silmeli miyiz yoksa dönüştürmeli miyiz?

> 💡 **Boxplot Okuma Kılavuzu:**
> - Kutunun alt kenarı → Q1 (25. yüzdelik)
> - Kutunun üst kenarı → Q3 (75. yüzdelik)
> - Kutunun içindeki çizgi → Medyan (Q2)
> - Bıyıklar → Q1 − 1.5×IQR ve Q3 + 1.5×IQR sınırları
> - Nokta/yıldız olarak görünen değerler → **Aykırı değerler**

> 🧠 **IQR Nedir?** Interquartile Range = Q3 − Q1. Aykırı değer tespitinde en yaygın kullanılan yöntemlerden biridir.

> ⚠️ **Dikkat:** `product_price` sütununda yüksek değerli siparişler (örn. elektronik) gerçek aykırı değer mi yoksa geçerli bir veri mi? Bağlamı göz ardı etmeyiniz!

</div>

````python
sayisal = ["product_price", "discount_rate", "delivery_time_days", "seller_rating", "product_rating"]

fig, axes = plt.subplots(1, len(sayisal), figsize=(16, 5))
fig.suptitle("Sayısal Sütunlarda Aykırı Değer Analizi (Box Plot)", fontsize=13, fontweight="bold", y=1.02)

for ax, col, color in zip(axes, sayisal, PASTEL_PALETTE):
    bp = ax.boxplot(
        df[col].dropna(),
        patch_artist=True,
        notch=False,
        boxprops=dict(facecolor=color, color="#888888"),
        medianprops=dict(color="#333333", linewidth=2),
        whiskerprops=dict(color="#888888"),
        capprops=dict(color="#888888"),
        flierprops=dict(marker="o", markerfacecolor=color, markeredgecolor="#888888",
                        markersize=4, alpha=0.5),
    )
    # Aykırı değer sayısını hesapla (IQR yöntemi)
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    aykiri = df[(df[col] < Q1 - 1.5 * IQR) | (df[col] > Q3 + 1.5 * IQR)][col].count()

    ax.set_title(f"{col}\n(Aykırı: {aykiri})", fontsize=9)
    ax.set_xticks([])
    ax.set_ylabel("Değer")

plt.tight_layout()
plt.show()
````

> 💡 **İpucu:** `unstack()` metodu, GroupBy sonucunu pivot tabloya dönüştürür. `fill_value=0` sayesinde sipariş olmayan kombinasyonlar NaN yerine 0 değeri alır.

````python
kategori_dagilim = df["product_category"].value_counts().sort_values(ascending=True)
renkler = PASTEL_PALETTE[:len(kategori_dagilim)]

fig, (ax_pie, ax_bar) = plt.subplots(1, 2, figsize=(14, 6))

# ── Sol: Pasta Grafiği ────────────────────────────────────────────────────────
wedges, texts, autotexts = ax_pie.pie(
    kategori_dagilim,
    labels=kategori_dagilim.index,
    colors=renkler,
    autopct="%1.1f%%",
    startangle=140,
    pctdistance=0.78,
    wedgeprops=dict(edgecolor="white", linewidth=1.5),
)
for at in autotexts:
    at.set_fontsize(10)
    at.set_color("#333333")
ax_pie.set_title("Kategori Bazında Sipariş Dağılımı\n(Pasta)", fontsize=13, fontweight="bold", pad=12)

# ── Sağ: Yatay Bar Grafiği ────────────────────────────────────────────────────
bars = ax_bar.barh(
    kategori_dagilim.index,
    kategori_dagilim.values,
    color=renkler,
    edgecolor="white",
    linewidth=0.8,
    height=0.6,
)

# Değer etiketleri
for bar in bars:
    w = bar.get_width()
    ax_bar.text(
        w + 80, bar.get_y() + bar.get_height() / 2,
        f"{w:,}", va="center", ha="left", fontsize=10, color="#444444"
    )

ax_bar.set_title("Kategori Bazında Sipariş Sayısı\n(Yatay Bar)", fontsize=13, fontweight="bold", pad=12)
ax_bar.set_xlabel("Sipariş Sayısı")
ax_bar.set_xlim(0, kategori_dagilim.max() * 1.15)
ax_bar.grid(axis="x", linestyle="--", linewidth=0.6, alpha=0.7)
ax_bar.spines[["top", "right"]].set_visible(False)

plt.suptitle("Ürün Kategorisi Dağılım Analizi", fontsize=15, fontweight="bold", y=1.02)
plt.tight_layout()
plt.show()
````

## 🌍 Bölüm 16 — **Plotly: Şehir Bazında Ortalama Sipariş Tutarı (İnteraktif Bar Chart)**

<div style="background: #FFF8F0; border-left: 5px solid #FF6000; padding: 15px 20px; border-radius: 0 8px 8px 0; margin: 10px 0;">

### 📋 **Senaryo**
Trendyol Strateji Direktörü, bölgesel gelir tablosunu bir sunumda kullanmak istiyor. Statik grafikler yeterli değil — grafiğin üzerine gelince şehir adını ve tam tutarı gösteren, zoom yapılabilen **interaktif** bir bar chart talep ediyor.

### 🎯 **Görev**
`plotly.express` kullanarak şehir bazında **ortalama sipariş tutarını** gösteren interaktif bir bar chart oluşturunuz.

**Yapmanız Gerekenler:**
1. Grup bazında ortalama hesaplayınız:
```python
sehir_ort = df.groupby("city")["product_price"].mean().round(2).reset_index()
sehir_ort.columns = ["Şehir", "Ortalama Tutar (TL)"]
sehir_ort = sehir_ort.sort_values("Ortalama Tutar (TL)", ascending=False)
```

2. `px.bar()` ile interaktif bar chart oluşturunuz:
```python
fig = px.bar(
    sehir_ort,
    x           = "Şehir",
    y           = "Ortalama Tutar (TL)",
    color       = "Şehir",
    text        = "Ortalama Tutar (TL)",
    color_discrete_sequence = px.colors.qualitative.Pastel,
    title       = "🏙️ Şehir Bazında Ortalama Sipariş Tutarı",
    template    = "plotly_white"
)
fig.update_traces(texttemplate="%{text:.2f} ₺", textposition="outside")
fig.update_layout(
    showlegend  = False,
    xaxis_title = "Şehir",
    yaxis_title = "Ortalama Tutar (TL)",
    title_font_size = 16
)
fig.show()
```

3. Grafik üzerinde **hover** yaparak değerleri inceleyiniz. Bir sütuna tıklayarak **zoom** yapınız.

4. Yorumlayınız:
   - Hangi şehirde ortalama sipariş tutarı en yüksek?
   - En düşük ile en yüksek şehir arasındaki fark ne kadar?

> 💡 **Plotly vs Matplotlib:** Plotly grafikleri HTML tabanlıdır — Jupyter'da interaktif çalışır, web uygulamalarına kolayca entegre edilebilir.

> 💡 **İpucu:** `fig.write_html("grafik.html")` ile grafiği bir HTML dosyası olarak dışa aktarabilirsiniz.

</div>

````python
sehir_ort = (
    df.groupby("city")["product_price"]
    .mean()
    .round(2)
    .sort_values(ascending=False)
    .reset_index()
    .rename(columns={"city": "Şehir", "product_price": "Ortalama Sipariş Tutarı (₺)"})
)

fig = px.bar(
    sehir_ort,
    x="Şehir",
    y="Ortalama Sipariş Tutarı (₺)",
    color="Şehir",
    color_discrete_sequence=PASTEL_PALETTE,
    text="Ortalama Sipariş Tutarı (₺)",
    title="Şehir Bazında Ortalama Sipariş Tutarı",
    labels={"Şehir": "Şehir", "Ortalama Sipariş Tutarı (₺)": "Ortalama Tutar (₺)"},
)

fig.update_traces(
    texttemplate="₺%{text:,.0f}",
    textposition="outside",
    marker_line_color="white",
    marker_line_width=1.2,
)

fig.update_layout(
    xaxis_title="Şehir",
    yaxis_title="Ortalama Tutar (₺)",
    yaxis_range=[0, sehir_ort["Ortalama Sipariş Tutarı (₺)"].max() * 1.15],
    showlegend=False,
    height=480,
    title_font_size=16,
    hoverlabel=dict(bgcolor="white", font_size=13),
)

fig.show()
````

## 🌀 **Bölüm 18 — Plotly: Kategori × Ödeme Yöntemi Sunburst Grafiği**

<div style="background: #FFF8F0; border-left: 5px solid #FF6000; padding: 15px 20px; border-radius: 0 8px 8px 0; margin: 10px 0;">

Trendyol Finans Ekibi, **hangi kategoride hangi ödeme yönteminin** ne kadar kullanıldığını hiyerarşik bir yapıda sunmak istiyor. Sunburst grafiği bu tür çok katmanlı kategorik verileri göstermede en etkili araçlardan biridir — hem payı hem de alt kırılımları aynı anda verir.

```python
sunburst_df = df.groupby(

fig = px.sunburst(
    sunburst_df,
    path    = ["product_category", "payment_type"],
    values  = "siparis_sayisi",
    color   = "product_category",
    color_discrete_sequence = px.colors.qualitative.Pastel,
```

```python
fig2 = px.box(
```

**Yorumlayınız:**
- Sunburst'te hangi kategori × ödeme yöntemi kombinasyonu en büyük dilimi oluşturuyor?
- Kapıda ödeme hangi kategorilerde daha yaygın? Bu mantıklı mı?
- Hangi kategoride satıcı puanının dağılımı en geniş? (IQR en büyük)

> 💡 **Sunburst Nasıl Okunur?** İç halka → üst level kategori (product_category), dış halka → alt level (payment_type). Bir dilime tıklayınca o kategoriye zoom yapabilirsiniz.
