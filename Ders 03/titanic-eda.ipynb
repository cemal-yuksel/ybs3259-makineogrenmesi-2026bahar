## 🎨 **Değişken Türlerinin Görsel Dağılımı**

Hangi tür değişkenlerimiz var ve dağılımları nasıl?

````python
# Değişken kategorilerinin görselleştirilmesi
from plotly.subplots import make_subplots
import plotly.graph_objects as go

# Veri hazırlama
variable_counts = {
    'Kategorik': len(categorical_features),
    'Sayısal (Sürekli)': len(continuous_features),
    'Sayısal (Kesikli)': len(discrete_features),
    'Hedef Değişken': 1
}

# Alt grafik yapısı oluşturma (1 satır, 2 sütun)
fig = make_subplots(
    rows=1, cols=2,
    specs=[[{"type": "pie"}, {"type": "bar"}]],
    subplot_titles=('Değişken Türleri Dağılımı', 'Değişken Sayıları')
)

# Pasta grafiği (sol)
fig.add_trace(
    go.Pie(
        labels=list(variable_counts.keys()),
        values=list(variable_counts.values()),
        hole=0.4,
        marker=dict(colors=['#FFB6C1', '#B0E0E6', '#98FB98', '#FFD700']),
        textinfo='label+percent',
        textfont=dict(size=13, color='#000000'),
        hovertemplate='<b>%{label}</b><br>Sayı: %{value}<br>Yüzde: %{percent}<extra></extra>'
    ),
    row=1, col=1
)

# Bar grafiği (sağ)
fig.add_trace(
    go.Bar(
        x=list(variable_counts.keys()),
        y=list(variable_counts.values()),
        marker=dict(
            color=['#FFB6C1', '#B0E0E6', '#98FB98', '#FFD700'],
            line=dict(color='#2c3e50', width=2)
        ),
        text=list(variable_counts.values()),
        textposition='outside',
        textfont=dict(size=14, color='#2c3e50'),
        hovertemplate='<b>%{x}</b><br>Sayı: %{y}<extra></extra>'
    ),
    row=1, col=2
)

# Grafik düzenlemesi
fig.update_layout(
    title=dict(
        text='Titanic Dataset Değişken Analizi',
        font=dict(size=22, color='#2c3e50', family='Arial Black'),
        x=0.5,
        xanchor='center'
    ),
    height=500,
    showlegend=False,
    paper_bgcolor='#fafafa',
    plot_bgcolor='#ffffff'
)

fig.update_xaxes(title_text="Değişken Türü", row=1, col=2, tickfont=dict(size=11))
fig.update_yaxes(title_text="Sayı", row=1, col=2, range=[0, max(variable_counts.values()) + 1])

fig.show()
````

## 🔍 **Eksik Veri Tespiti**

İlk olarak, veri setimizdeki eksik değerleri tespit edelim.

````python
# Eksik veri analizi
missing_data = pd.DataFrame({
    'Değişken': df.columns,
    'Eksik Sayı': df.isnull().sum().values,
    'Eksik Yüzde': (df.isnull().sum() / len(df) * 100).round(2).values,
    'Veri Tipi': df.dtypes.values,
    'Toplam Değer': len(df)
})

# Sadece eksik verisi olanları filtrele
missing_data = missing_data[missing_data['Eksik Sayı'] > 0].sort_values('Eksik Yüzde', ascending=False)

print("🕳️ EKSİK VERİ ANALİZİ")
print("="*90)
print(f"\n📊 Toplam Değişken Sayısı: {len(df.columns)}")
print(f"⚠️  Eksik Verisi Olan Değişken: {len(missing_data)}")
print(f"✅ Tam Verisi Olan Değişken: {len(df.columns) - len(missing_data)}")
print("\n" + "="*90)
print("📋 EKSİK VERİ DETAYLARI")
print("="*90)
print(missing_data.to_string(index=False))
print("="*90)
````

## 📊 **Eksik Veri Görselleştirmesi**

Eksik verileri görselleştirerek daha iyi anlayalım.

````python
# Eksik veri yüzdeleri
missing_percentage = (df.isnull().sum() / len(df) * 100).sort_values(ascending=False)
missing_percentage = missing_percentage[missing_percentage > 0]

# Bar chart ile görselleştirme
fig = go.Figure()

fig.add_trace(go.Bar(
    x=missing_percentage.values,
    y=missing_percentage.index,
    orientation='h',
    marker=dict(
        color=missing_percentage.values,
        colorscale='Reds',
        showscale=True,
        colorbar=dict(title=dict(text='Eksik %', font=dict(color='#2c3e50'))),
        line=dict(color='#c0392b', width=2)
    ),
    text=[f'{val:.1f}%' for val in missing_percentage.values],
    textposition='outside',
    textfont=dict(size=13, color='#2c3e50', family='Arial Black'),
    hovertemplate='<b>%{y}</b><br>Eksik Yüzde: %{x:.2f}%<br>Eksik Sayı: %{text}<extra></extra>',
    customdata=[(df[col].isnull().sum()) for col in missing_percentage.index]
))

fig.update_layout(
    title=dict(
        text='Değişkenlere Göre Eksik Veri Yüzdeleri',
        font=dict(size=22, color='#2c3e50', family='Arial Black'),
        x=0.5,
        xanchor='center'
    ),
    xaxis=dict(
        title=dict(text='Eksik Veri Yüzdesi (%)', font=dict(size=14, color='#2c3e50')),
        gridcolor='#ecf0f1'
    ),
    yaxis=dict(
        title=dict(text='Değişkenler', font=dict(size=14, color='#2c3e50')),
        tickfont=dict(size=12, color='#2c3e50')
    ),
    height=500,
    paper_bgcolor='#fafafa',
    plot_bgcolor='#ffffff',
    showlegend=False
)

fig.show()

# Özet istatistikler
print("\n📌 EKSİK VERİ ÖZETİ:")
print("="*70)
print(f"   • En fazla eksik: {missing_percentage.index[0]} ({missing_percentage.values[0]:.1f}%)")
print(f"   • En az eksik: {missing_percentage.index[-1]} ({missing_percentage.values[-1]:.1f}%)")
print(f"   • Ortalama eksik: {missing_percentage.mean():.1f}%")
print("="*70)
````

### 👶 Yaş: `Age`

**Sürekli değişken** — Yolcuların yaşları <code>(0.42 ile 80 arasında)</code>


````python
# Yaş istatistikleri (NaN hariç)
age_data = df['Age'].dropna()

# Görselleştirme
fig = make_subplots(
    rows=2, cols=2,
    subplot_titles=('Yaş Dağılımı - Histogram', 'Yaş Kutu Grafiği (Box Plot)',
                    'Yaş Yoğunluk Grafiği (KDE)', 'Yaş İstatistikleri'),
    specs=[[{"type": "histogram"}, {"type": "box"}],
           [{"type": "scatter"}, {"type": "table"}]],
    vertical_spacing=0.12,
    horizontal_spacing=0.1
)

# 1. Histogram
fig.add_trace(
    go.Histogram(
        x=age_data,
        nbinsx=30,
        marker=dict(color='#3498db', line=dict(color='#2c3e50', width=1)),
        name='Histogram',
        hovertemplate='Yaş: %{x}<br>Frekans: %{y}<extra></extra>'
    ),
    row=1, col=1
)

# 2. Box Plot
fig.add_trace(
    go.Box(
        y=age_data,
        marker=dict(color='#e74c3c'),
        name='Box Plot',
        boxmean='sd',  # Ortalama ve standart sapma göster
        hovertemplate='Yaş: %{y}<extra></extra>'
    ),
    row=1, col=2
)

# 3. KDE (Yoğunluk) - Matplotlib ile hesaplayıp Plotly'ye aktarıyoruz
from scipy import stats
density = stats.gaussian_kde(age_data)
xs = np.linspace(age_data.min(), age_data.max(), 200)
fig.add_trace(
    go.Scatter(
        x=xs,
        y=density(xs),
        mode='lines',
        fill='tozeroy',
        line=dict(color='#2ecc71', width=3),
        name='KDE',
        hovertemplate='Yaş: %{x:.1f}<br>Yoğunluk: %{y:.4f}<extra></extra>'
    ),
    row=2, col=1
)

# 4. İstatistik tablosu
stats_data = [
    ['Ortalama (Mean)', f'{age_data.mean():.2f}'],
    ['Medyan', f'{age_data.median():.2f}'],
    ['Mod', f'{age_data.mode()[0]:.2f}'],
    ['Std Sapma', f'{age_data.std():.2f}'],
    ['Minimum', f'{age_data.min():.2f}'],
    ['Maksimum', f'{age_data.max():.2f}'],
    ['Q1 (25%)', f'{age_data.quantile(0.25):.2f}'],
    ['Q3 (75%)', f'{age_data.quantile(0.75):.2f}'],
    ['IQR', f'{age_data.quantile(0.75) - age_data.quantile(0.25):.2f}']
]

fig.add_trace(
    go.Table(
        header=dict(
            values=['<b>Metrik</b>', '<b>Değer</b>'],
            fill_color='#3498db',
            align='left',
            font=dict(color='white', size=13, family='Arial Black')
        ),
        cells=dict(
            values=list(zip(*stats_data)),
            fill_color=['#ecf0f1', '#ffffff'],
            align='left',
            font=dict(size=12)
        )
    ),
    row=2, col=2
)

fig.update_layout(
    title=dict(
        text='Yaş Değişkeni: Kapsamlı Analiz',
        font=dict(size=20, color='#2c3e50', family='Arial Black'),
        x=0.5,
        xanchor='center'
    ),
    height=800,
    showlegend=False,
    paper_bgcolor='#fafafa'
)

fig.update_xaxes(title=dict(text='Yaş', font=dict(size=11)), row=1, col=1)
fig.update_yaxes(title=dict(text='Frekans', font=dict(size=11)), row=1, col=1)
fig.update_yaxes(title=dict(text='Yaş', font=dict(size=11)), row=1, col=2)
fig.update_xaxes(title=dict(text='Yaş', font=dict(size=11)), row=2, col=1)
fig.update_yaxes(title=dict(text='Yoğunluk', font=dict(size=11)), row=2, col=1)

fig.show()
````

### 💰 **Bilet Ücreti: Fare**

**Sürekli değişken** - Yolcuların ödediği bilet ücretleri (£0 ile £512 arasında)

````python
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import numpy as np

fare_data = df['Fare'].dropna()

q1   = fare_data.quantile(0.25)
q3   = fare_data.quantile(0.75)
iqr  = q3 - q1
mean = fare_data.mean()
med  = fare_data.median()
skew = fare_data.skew()
outlier_count = int((fare_data > q3 + 1.5 * iqr).sum())

# --- Plotly: 2x2 subplot grid ---
fig = make_subplots(
    rows=2, cols=2,
    subplot_titles=(
        "Histogram — Frekans Dağılımı",
        "Box Plot — Aykırı Degerler",
        "Histogram — Log Olcekli",
        "Ozet Istatistikler"
    ),
    specs=[
        [{"type": "xy"}, {"type": "xy"}],
        [{"type": "xy"}, {"type": "table"}]
    ],
    horizontal_spacing=0.12,
    vertical_spacing=0.18
)

# 1. Histogram
fig.add_trace(
    go.Histogram(
        x=fare_data,
        nbinsx=60,
        name="Fare",
        marker_color="rgba(155, 89, 182, 0.75)",
        marker_line=dict(color="#2c3e50", width=0.5),
        showlegend=False
    ),
    row=1, col=1
)
for val, label, color in [(mean, f"Ortalama: £{mean:.2f}", "#e74c3c"),
                            (med,  f"Medyan: £{med:.2f}",    "#2ecc71")]:
    fig.add_vline(x=val, line_dash="dash", line_color=color,
                  annotation_text=label,
                  annotation_position="top right",
                  annotation_font_size=11,
                  row=1, col=1)

# 2. Box Plot
fig.add_trace(
    go.Box(
        y=fare_data,
        name="Fare",
        marker_color="#f39c12",
        line_color="#2c3e50",
        boxmean=True,
        showlegend=False
    ),
    row=1, col=2
)

# 3. Log-scale Histogram
fig.add_trace(
    go.Histogram(
        x=np.log1p(fare_data[fare_data > 0]),
        nbinsx=50,
        name="log(Fare+1)",
        marker_color="rgba(26, 188, 156, 0.75)",
        marker_line=dict(color="#2c3e50", width=0.5),
        showlegend=False
    ),
    row=2, col=1
)

# 4. Istatistik tablosu
stats_labels = [
    "Ortalama (Mean)", "Medyan", "Mod",
    "Std Sapma", "Minimum", "Maksimum",
    "Q1 (25%)", "Q2 (50%)", "Q3 (75%)",
    "IQR", "Carpiklik (Skewness)", "Aykiri Deger Sayisi"
]
stats_values = [
    f"£{mean:.2f}", f"£{med:.2f}", f"£{fare_data.mode()[0]:.2f}",
    f"£{fare_data.std():.2f}", f"£{fare_data.min():.2f}", f"£{fare_data.max():.2f}",
    f"£{q1:.2f}", f"£{med:.2f}", f"£{q3:.2f}",
    f"£{iqr:.2f}", f"{skew:.2f}", str(outlier_count)
]

fig.add_trace(
    go.Table(
        header=dict(
            values=["Metrik", "Deger"],
            fill_color="#302b63",
            font=dict(color="white", size=13),
            align="left",
            height=30
        ),
        cells=dict(
            values=[stats_labels, stats_values],
            fill_color=[["#1a1a2e", "#16213e"] * 6, ["#1a1a2e", "#16213e"] * 6],
            font=dict(color=["#80deea", "#f9ca24"], size=12),
            align="left",
            height=26
        )
    ),
    row=2, col=2
)

fig.update_layout(
    title=dict(
        text="Bilet Ucreti (Fare) — Dagilim Analizi",
        x=0.5, xanchor="center",
        font=dict(size=18, color="#2c3e50")
    ),
    height=700,
    paper_bgcolor="white",
    plot_bgcolor="#f8f9fa",
    font=dict(family="Arial", size=12)
)

fig.update_xaxes(title_text="Bilet Ucreti (£)", row=1, col=1, showgrid=True, gridcolor="#e0e0e0")
fig.update_yaxes(title_text="Frekans",          row=1, col=1, showgrid=True, gridcolor="#e0e0e0")
fig.update_xaxes(title_text="log(Fare + 1)",    row=2, col=1, showgrid=True, gridcolor="#e0e0e0")
fig.update_yaxes(title_text="Frekans",          row=2, col=1, showgrid=True, gridcolor="#e0e0e0")
fig.update_yaxes(title_text="Bilet Ucreti (£)", row=1, col=2, showgrid=True, gridcolor="#e0e0e0")

fig.show()
````

## 🎫 **Survived vs Pclass: Sosyal Sınıf ve Hayatta Kalma**

**Hipotez:** Üst sınıf yolcuların hayatta kalma şansı daha yüksektir (can simitleri ve botlara öncelikli erişim).

````python

# Cross-tabulation: Pclass vs Survived
ct_pclass     = pd.crosstab(df['Pclass'], df['Survived'], margins=True)
ct_pclass_pct = pd.crosstab(df['Pclass'], df['Survived'], normalize='index') * 100

sınıflar     = ['1. Sınıf', '2. Sınıf', '3. Sınıf']
oldu_renk    = '#e74c3c'
hayatta_renk = '#2ecc71'

# ── FİGÜR 1: Grouped Bar — Sayılar + Yüzdeler ───────────────────────────
fig1 = make_subplots(
    rows=1, cols=2,
    subplot_titles=('Hayatta Kalma Sayıları', 'Hayatta Kalma Oranları (%)'),
    horizontal_spacing=0.16
)

for survived, color, name, show in [
    (0, oldu_renk,    'Öldü',    True),
    (1, hayatta_renk, 'Hayatta', True)
]:
    counts = [int(ct_pclass.loc[i, survived]) for i in [1, 2, 3]]
    pcts   = [ct_pclass_pct.loc[i, survived]  for i in [1, 2, 3]]

    fig1.add_trace(
        go.Bar(x=sınıflar, y=counts,
               name=name, marker_color=color,
               marker_line=dict(color='#2c3e50', width=1.5),
               text=counts, textposition='outside',
               textfont=dict(size=13, color='#2c3e50'),
               legendgroup=name, showlegend=show),
        row=1, col=1
    )
    fig1.add_trace(
        go.Bar(x=sınıflar, y=pcts,
               name=name, marker_color=color,
               marker_line=dict(color='#2c3e50', width=1.5),
               text=[f'{p:.1f}%' for p in pcts], textposition='outside',
               textfont=dict(size=13, color='#2c3e50'),
               legendgroup=name, showlegend=False),
        row=1, col=2
    )

# Subplot başlıklarını ana başlıktan ayrı tutmak için aşağı kaydır
fig1.update_annotations(font_size=14, yshift=-8)

fig1.update_layout(
    title=dict(
        text='Bilet Sınıfı ve Hayatta Kalma',
        x=0.5, xanchor='center', y=0.97,
        font=dict(size=20, color='#2c3e50', family='Arial Black')
    ),
    barmode='group',
    height=520,
    paper_bgcolor='white',
    plot_bgcolor='#f8f9fa',
    font=dict(size=13, family='Arial'),
    legend=dict(
        title='Durum', font=dict(size=13),
        bgcolor='rgba(255,255,255,0.85)',
        bordercolor='#aaa', borderwidth=1
    ),
    margin=dict(t=130, b=70, l=70, r=40)
)
fig1.update_yaxes(title_text='Kişi Sayısı', title_font_size=13, row=1, col=1,
                  showgrid=True, gridcolor='#ddd', gridwidth=1)
fig1.update_yaxes(title_text='Oran (%)',    title_font_size=13, row=1, col=2,
                  showgrid=True, gridcolor='#ddd', gridwidth=1, range=[0, 110])
fig1.update_xaxes(tickfont=dict(size=13), row=1, col=1)
fig1.update_xaxes(tickfont=dict(size=13), row=1, col=2)
fig1.show()

# ── FİGÜR 2: Yığılmış %100 Bar ──────────────────────────────────────────
fig2 = go.Figure()

for survived, color, name in [
    (0, oldu_renk,    'Öldü'),
    (1, hayatta_renk, 'Hayatta')
]:
    pcts = [ct_pclass_pct.loc[i, survived] for i in [1, 2, 3]]
    fig2.add_trace(
        go.Bar(
            x=sınıflar, y=pcts,
            name=name, marker_color=color,
            marker_line=dict(color='#2c3e50', width=1.5),
            text=[f'{p:.1f}%' for p in pcts], textposition='inside',
            textfont=dict(size=14, color='white')
        )
    )

fig2.update_layout(
    title=dict(
        text='Bilet Sınıfına Göre Hayatta Kalma — Yığılmış %100',
        x=0.5, xanchor='center', y=0.95,
        font=dict(size=20, color='#2c3e50', family='Arial Black')
    ),
    barmode='stack',
    height=480,
    paper_bgcolor='white',
    plot_bgcolor='#f8f9fa',
    font=dict(size=13, family='Arial'),
    legend=dict(
        title='Durum', font=dict(size=13),
        bgcolor='rgba(255,255,255,0.85)',
        bordercolor='#aaa', borderwidth=1
    ),
    yaxis=dict(
        title='Oran (%)', title_font_size=13,
        showgrid=True, gridcolor='#ddd', gridwidth=1
    ),
    xaxis=dict(
        title='Bilet Sınıfı', title_font_size=13,
        tickfont=dict(size=14)
    ),
    margin=dict(t=100, b=70, l=70, r=40)
)
fig2.show()

# ── Özet ──────────────────────────────────────────────────────────────────
print("BİLET SINIFI — HAYATTA KALMA ORANLARI")
print("-" * 42)
for i in [1, 2, 3]:
    oran = ct_pclass_pct.loc[i, 1]
    print(f"  {i}. Sınıf  ->  Hayatta Kalma: {oran:.1f}%")
print("-" * 42)
````

## 👶 **Survived vs Age: Yaş ve Hayatta Kalma**

## 💰 **Survived vs Fare: Bilet Ücreti ve Hayatta Kalma**

**Ekonomik faktör analizi** - Daha pahalı bilet alanlar daha fazla hayatta kaldı mı?

````python

# Fare istatistikleri
fare_stats = df.groupby('Survived')['Fare'].describe()

oldu_renk    = '#e74c3c'
hayatta_renk = '#2ecc71'

# Fare grupları
fare_bins   = [0, 10, 20, 50, 100, 520]
fare_labels = ['£0-10', '£10-20', '£20-50', '£50-100', '£100+']
df_temp = df.copy()
df_temp['Fare_Group'] = pd.cut(df_temp['Fare'], bins=fare_bins, labels=fare_labels, include_lowest=True)
fare_group_surv = (df_temp.groupby('Fare_Group', observed=True)['Survived']
                   .agg(['sum', 'count', 'mean']).reset_index())
fare_group_surv['Oran'] = fare_group_surv['mean'] * 100

# ── FİGÜR 1: Histogram + Box Plot ───────────────────────────────────────
fig1 = make_subplots(
    rows=1, cols=2,
    subplot_titles=('Bilet Ücreti Dağılımı (Log Ölçek)', 'Bilet Ücreti Box Plot'),
    horizontal_spacing=0.14
)

for survived, color, name, show in [
    (0, oldu_renk,    'Öldü',    True),
    (1, hayatta_renk, 'Hayatta', True)
]:
    data = df[df['Survived'] == survived]['Fare']
    data_pos = data[data > 0]

    fig1.add_trace(
        go.Histogram(x=np.log1p(data_pos), nbinsx=35,
                     name=name, marker_color=color, opacity=0.70,
                     marker_line=dict(color='#2c3e50', width=0.8),
                     legendgroup=name, showlegend=show),
        row=1, col=1
    )
    fig1.add_trace(
        go.Box(y=data, name=name, marker_color=color,
               line=dict(color='#2c3e50', width=1.5),
               boxmean=True,
               legendgroup=name, showlegend=False),
        row=1, col=2
    )

fig1.update_annotations(font_size=14, yshift=-8)
fig1.update_layout(
    title=dict(
        text='Bilet Ücreti (Fare) ve Hayatta Kalma',
        x=0.5, xanchor='center', y=0.97,
        font=dict(size=20, color='#2c3e50', family='Arial Black')
    ),
    barmode='overlay',
    height=520,
    paper_bgcolor='white',
    plot_bgcolor='#f8f9fa',
    font=dict(size=13, family='Arial'),
    legend=dict(title='Durum', font=dict(size=13),
                bgcolor='rgba(255,255,255,0.85)',
                bordercolor='#aaa', borderwidth=1),
    margin=dict(t=130, b=70, l=70, r=40)
)
fig1.update_xaxes(title_text='ln(Fare + 1)', title_font_size=13,
                  tickfont=dict(size=12), row=1, col=1)
fig1.update_yaxes(title_text='Frekans', title_font_size=13,
                  showgrid=True, gridcolor='#ddd', row=1, col=1)
fig1.update_xaxes(tickfont=dict(size=13), row=1, col=2)
fig1.update_yaxes(title_text='Bilet Ücreti (£)', title_font_size=13,
                  showgrid=True, gridcolor='#ddd', row=1, col=2)
fig1.show()

# ── FİGÜR 2: Ücret Grubuna Göre Hayatta Kalma Oranı ─────────────────────
grup_renkleri = ['#e74c3c', '#e67e22', '#f39c12', '#2ecc71', '#27ae60']

fig2 = go.Figure()
fig2.add_trace(
    go.Bar(
        x=fare_labels,
        y=fare_group_surv['Oran'],
        marker_color=grup_renkleri,
        marker_line=dict(color='#2c3e50', width=1.5),
        text=[f'{v:.1f}%' for v in fare_group_surv['Oran']],
        textposition='outside',
        textfont=dict(size=13, color='#2c3e50'),
        showlegend=False
    )
)
fig2.update_layout(
    title=dict(
        text='Ücret Grubuna Göre Hayatta Kalma Oranı',
        x=0.5, xanchor='center', y=0.95,
        font=dict(size=20, color='#2c3e50', family='Arial Black')
    ),
    height=460,
    paper_bgcolor='white',
    plot_bgcolor='#f8f9fa',
    font=dict(size=13, family='Arial'),
    xaxis=dict(title='Ücret Grubu', title_font_size=13, tickfont=dict(size=13)),
    yaxis=dict(title='Hayatta Kalma Oranı (%)', title_font_size=13,
               showgrid=True, gridcolor='#ddd', range=[0, 100]),
    margin=dict(t=100, b=70, l=70, r=40)
)
fig2.show()

# ── FİGÜR 3: İstatistik Tablosu ──────────────────────────────────────────
satirlar = ['Ortalama (£)', 'Medyan (£)', 'Std Sapma (£)', 'Min (£)', 'Maks (£)']
oldu_vals    = [f"{fare_stats.loc[0, 'mean']:.2f}",
                f"{fare_stats.loc[0, '50%']:.2f}",
                f"{fare_stats.loc[0, 'std']:.2f}",
                f"{fare_stats.loc[0, 'min']:.2f}",
                f"{fare_stats.loc[0, 'max']:.2f}"]
hayatta_vals = [f"{fare_stats.loc[1, 'mean']:.2f}",
                f"{fare_stats.loc[1, '50%']:.2f}",
                f"{fare_stats.loc[1, 'std']:.2f}",
                f"{fare_stats.loc[1, 'min']:.2f}",
                f"{fare_stats.loc[1, 'max']:.2f}"]
fark_vals    = [f"{fare_stats.loc[1, 'mean']  - fare_stats.loc[0, 'mean']:.2f}",
                f"{fare_stats.loc[1, '50%']   - fare_stats.loc[0, '50%']:.2f}",
                f"{fare_stats.loc[1, 'std']   - fare_stats.loc[0, 'std']:.2f}",
                "-", "-"]

fig3 = go.Figure(go.Table(
    header=dict(
        values=['<b>İstatistik</b>', '<b>Öldü</b>', '<b>Hayatta</b>', '<b>Fark</b>'],
        fill_color=['#2c3e50', oldu_renk, hayatta_renk, '#3498db'],
        font=dict(color='white', size=14, family='Arial Black'),
        align='center', height=36
    ),
    cells=dict(
        values=[satirlar, oldu_vals, hayatta_vals, fark_vals],
        fill_color=[['#f8f9fa'] * 5,
                    ['#fdecea'] * 5,
                    ['#eafaf1'] * 5,
                    ['#ebf5fb'] * 5],
        font=dict(color='#2c3e50', size=13, family='Arial'),
        align='center', height=32
    )
))
fig3.update_layout(
    title=dict(
        text='Bilet Ücreti — Grup Karşılaştırması',
        x=0.5, xanchor='center', y=0.95,
        font=dict(size=18, color='#2c3e50', family='Arial Black')
    ),
    height=300,
    paper_bgcolor='white',
    margin=dict(t=90, b=30, l=30, r=30)
)
fig3.show()
````

## ⚓ **Survived vs Embarked: Biniş Limanı ve Hayatta Kalma**

**Coğrafi faktör** - Hangi limandan binen yolcuların şansı daha yüksekti?

````python

# Cross-tabulation: Embarked vs Survived (NaN'ları çıkar)
embarked_survived = df[df['Embarked'].notna()].copy()
ct_embarked     = pd.crosstab(embarked_survived['Embarked'], embarked_survived['Survived'], margins=True)
ct_embarked_pct = pd.crosstab(embarked_survived['Embarked'], embarked_survived['Survived'], normalize='index') * 100

port_names  = {'C': 'Cherbourg', 'Q': 'Queenstown', 'S': 'Southampton'}
port_labels = [f'{port_names[p]}<br>({p})' for p in ['C', 'Q', 'S']]
oldu_renk    = '#e74c3c'
hayatta_renk = '#2ecc71'

# ── FİGÜR 1: Grouped Bar — Sayılar + Oranlar ─────────────────────────────
fig1 = make_subplots(
    rows=1, cols=2,
    subplot_titles=('Limana Göre Hayatta Kalma Sayıları', 'Limana Göre Hayatta Kalma Oranları (%)'),
    horizontal_spacing=0.16
)

for survived, color, name, show in [
    (0, oldu_renk,    'Öldü',    True),
    (1, hayatta_renk, 'Hayatta', True)
]:
    counts = [int(ct_embarked.loc[p, survived]) for p in ['C', 'Q', 'S']]
    pcts   = [ct_embarked_pct.loc[p, survived]  for p in ['C', 'Q', 'S']]

    fig1.add_trace(
        go.Bar(x=port_labels, y=counts,
               name=name, marker_color=color,
               marker_line=dict(color='#2c3e50', width=1.5),
               text=counts, textposition='outside',
               textfont=dict(size=13, color='#2c3e50'),
               legendgroup=name, showlegend=show),
        row=1, col=1
    )
    fig1.add_trace(
        go.Bar(x=port_labels, y=pcts,
               name=name, marker_color=color,
               marker_line=dict(color='#2c3e50', width=1.5),
               text=[f'{p:.1f}%' for p in pcts], textposition='outside',
               textfont=dict(size=13, color='#2c3e50'),
               legendgroup=name, showlegend=False),
        row=1, col=2
    )

fig1.update_annotations(font_size=14, yshift=-8)
fig1.update_layout(
    title=dict(
        text='Biniş Limanı ve Hayatta Kalma',
        x=0.5, xanchor='center', y=0.97,
        font=dict(size=20, color='#2c3e50', family='Arial Black')
    ),
    barmode='group',
    height=520,
    paper_bgcolor='white',
    plot_bgcolor='#f8f9fa',
    font=dict(size=13, family='Arial'),
    legend=dict(title='Durum', font=dict(size=13),
                bgcolor='rgba(255,255,255,0.85)',
                bordercolor='#aaa', borderwidth=1),
    margin=dict(t=130, b=80, l=70, r=40)
)
fig1.update_yaxes(title_text='Kişi Sayısı', title_font_size=13, row=1, col=1,
                  showgrid=True, gridcolor='#ddd')
fig1.update_yaxes(title_text='Oran (%)',    title_font_size=13, row=1, col=2,
                  showgrid=True, gridcolor='#ddd', range=[0, 110])
fig1.update_xaxes(tickfont=dict(size=12), row=1, col=1)
fig1.update_xaxes(tickfont=dict(size=12), row=1, col=2)
fig1.show()

# ── FİGÜR 2: Yığılmış %100 Bar ──────────────────────────────────────────
fig2 = go.Figure()

for survived, color, name in [
    (0, oldu_renk,    'Öldü'),
    (1, hayatta_renk, 'Hayatta')
]:
    pcts = [ct_embarked_pct.loc[p, survived] for p in ['C', 'Q', 'S']]
    fig2.add_trace(
        go.Bar(x=port_labels, y=pcts,
               name=name, marker_color=color,
               marker_line=dict(color='#2c3e50', width=1.5),
               text=[f'{p:.1f}%' for p in pcts], textposition='inside',
               textfont=dict(size=14, color='white'))
    )

fig2.update_layout(
    title=dict(
        text='Biniş Limanına Göre Hayatta Kalma — Yığılmış %100',
        x=0.5, xanchor='center', y=0.95,
        font=dict(size=20, color='#2c3e50', family='Arial Black')
    ),
    barmode='stack',
    height=480,
    paper_bgcolor='white',
    plot_bgcolor='#f8f9fa',
    font=dict(size=13, family='Arial'),
    legend=dict(title='Durum', font=dict(size=13),
                bgcolor='rgba(255,255,255,0.85)',
                bordercolor='#aaa', borderwidth=1),
    yaxis=dict(title='Oran (%)', title_font_size=13,
               showgrid=True, gridcolor='#ddd'),
    xaxis=dict(title='Biniş Limanı', title_font_size=13, tickfont=dict(size=13)),
    margin=dict(t=100, b=80, l=70, r=40)
)
fig2.show()

# ── FİGÜR 3: Liman × Sınıf Dağılımı ─────────────────────────────────────
embarked_class     = embarked_survived.groupby(['Embarked', 'Pclass']).size().unstack(fill_value=0)
embarked_class_pct = embarked_class.div(embarked_class.sum(axis=1), axis=0) * 100

sinif_renkleri = {'1': '#3498db', '2': '#f39c12', '3': '#e74c3c'}

fig3 = go.Figure()
for pclass in [1, 2, 3]:
    pcts = [embarked_class_pct.loc[p, pclass] for p in ['C', 'Q', 'S']]
    fig3.add_trace(
        go.Bar(x=port_labels, y=pcts,
               name=f'{pclass}. Sınıf',
               marker_color=sinif_renkleri[str(pclass)],
               marker_line=dict(color='#2c3e50', width=1.2),
               text=[f'{v:.0f}%' for v in pcts], textposition='inside',
               textfont=dict(size=13, color='white'))
    )

fig3.update_layout(
    title=dict(
        text='Biniş Limanı × Bilet Sınıfı Dağılımı (%)',
        x=0.5, xanchor='center', y=0.95,
        font=dict(size=20, color='#2c3e50', family='Arial Black')
    ),
    barmode='stack',
    height=480,
    paper_bgcolor='white',
    plot_bgcolor='#f8f9fa',
    font=dict(size=13, family='Arial'),
    legend=dict(title='Sınıf', font=dict(size=13),
                bgcolor='rgba(255,255,255,0.85)',
                bordercolor='#aaa', borderwidth=1),
    yaxis=dict(title='Oran (%)', title_font_size=13,
               showgrid=True, gridcolor='#ddd'),
    xaxis=dict(title='Biniş Limanı', title_font_size=13, tickfont=dict(size=13)),
    margin=dict(t=100, b=80, l=70, r=40)
)
fig3.show()

# ── Özet ──────────────────────────────────────────────────────────────────
print("BINIS LIMANI — HAYATTA KALMA ORANLARI")
print("-" * 50)
for p in ['C', 'Q', 'S']:
    oran = ct_embarked_pct.loc[p, 1]
    print(f"  {port_names[p]:15s} ({p})  ->  Hayatta Kalma: {oran:.1f}%")
print("-" * 50)
print("BINIS LIMANI x SINIF DAGILIMI (%):")
print(embarked_class_pct.round(1).to_string())
print("-" * 50)
````

## 🌟 **Feature Importance: Hangi Değişken Ne Kadar Önemli?**

Basit bir **Decision Tree** modeli ile değişken önemlerini (feature importance) bulalım.

````python

from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import LabelEncoder

# Veri hazırlama
df_model = df[['Survived', 'Pclass', 'Sex', 'Age', 'SibSp', 'Parch', 'Fare', 'Embarked']].dropna().copy()

le_sex      = LabelEncoder()
le_embarked = LabelEncoder()
df_model['Sex_encoded']      = le_sex.fit_transform(df_model['Sex'])
df_model['Embarked_encoded'] = le_embarked.fit_transform(df_model['Embarked'])

features = ['Pclass', 'Sex_encoded', 'Age', 'SibSp', 'Parch', 'Fare', 'Embarked_encoded']
feat_labels = ['Pclass', 'Cinsiyet', 'Yaş', 'SibSp', 'Parch', 'Ücret', 'Liman']

X = df_model[features]
y = df_model['Survived']

dt_model = DecisionTreeClassifier(max_depth=5, random_state=42)
dt_model.fit(X, y)

importance_df = pd.DataFrame({
    'Değişken': feat_labels,
    'Önem': dt_model.feature_importances_
}).sort_values('Önem', ascending=True)

# Renk skalası: düşük → yüksek
bar_palette = [
    "#e74c3c", "#e67e22", "#f1c40f",
    "#2ecc71", "#27ae60", "#1abc9c", "#16a085"
]
colors = bar_palette[:len(importance_df)]

# ── FİGÜR 1: Yatay Bar Chart ─────────────────────────────────────────────
fig1 = go.Figure(go.Bar(
    x=importance_df['Önem'],
    y=importance_df['Değişken'],
    orientation='h',
    marker=dict(
        color=importance_df['Önem'],
        colorscale=[[0, "#e74c3c"], [0.5, "#f1c40f"], [1, "#27ae60"]],
        showscale=True,
        colorbar=dict(
            title=dict(text="Önem", font=dict(size=12)),
            thickness=14,
            tickfont=dict(size=11)
        ),
        line=dict(color="#2c3e50", width=1)
    ),
    text=[f"{v:.4f}" for v in importance_df['Önem']],
    textposition='outside',
    textfont=dict(size=13, color="#2c3e50")
))

fig1.update_layout(
    title=dict(
        text="Değişken Önem Sıralaması — Decision Tree",
        x=0.5, xanchor="center", y=0.95,
        font=dict(size=20, color="#2c3e50", family="Arial Black")
    ),
    height=420,
    paper_bgcolor="white",
    plot_bgcolor="#f8f9fa",
    font=dict(size=13, family="Arial"),
    xaxis=dict(
        title="Önem Skoru",
        title_font_size=13,
        tickfont=dict(size=12),
        gridcolor="#ddd", showgrid=True,
        range=[0, importance_df['Önem'].max() * 1.18]
    ),
    yaxis=dict(tickfont=dict(size=13)),
    showlegend=False,
    margin=dict(t=100, b=70, l=90, r=80)
)
fig1.show()

# ── FİGÜR 2: Donut Chart ─────────────────────────────────────────────────
imp_sorted = importance_df.sort_values('Önem', ascending=False)

fig2 = go.Figure(go.Pie(
    labels=imp_sorted['Değişken'],
    values=imp_sorted['Önem'],
    hole=0.45,
    marker=dict(
        colors=["#27ae60", "#1abc9c", "#f1c40f", "#e67e22",
                "#e74c3c", "#8e44ad", "#3498db"],
        line=dict(color="white", width=2)
    ),
    textinfo='label+percent',
    textfont=dict(size=13),
    hovertemplate="<b>%{label}</b><br>Önem: %{value:.4f}<br>Oran: %{percent}<extra></extra>"
))

fig2.add_annotation(
    text="<b>Önem<br>Dağılımı</b>",
    x=0.5, y=0.5,
    font=dict(size=14, color="#2c3e50"),
    showarrow=False
)

fig2.update_layout(
    title=dict(
        text="Değişken Önem Dağılımı (%)",
        x=0.5, xanchor="center", y=0.95,
        font=dict(size=20, color="#2c3e50", family="Arial Black")
    ),
    height=460,
    paper_bgcolor="white",
    font=dict(size=13, family="Arial"),
    legend=dict(font=dict(size=12), bgcolor="rgba(255,255,255,0.85)", borderwidth=1),
    margin=dict(t=100, b=60, l=60, r=60)
)
fig2.show()

# ── Özet ──────────────────────────────────────────────────────────────────
print("FEATURE IMPORTANCE — ÖZET")
print("-" * 50)
for _, row in importance_df.sort_values('Önem', ascending=False).iterrows():
    bar = "█" * int(row['Önem'] * 80)
    print(f"  {row['Değişken']:10s}  {row['Önem']:.4f}  {bar}")
print("-" * 50)
top3_toplam = importance_df.sort_values('Önem', ascending=False).head(3)['Önem'].sum()
print(f"  İlk 3 değişkeni toplam önemi: %{top3_toplam*100:.1f}")
print()
print("  NOT: Skorlar Decision Tree (max_depth=5) modeline göre hesaplanmıştır.")
print("       Farklı modellerde sıralama değişebilir.")
print("-" * 50)
````
